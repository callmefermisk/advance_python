```python
# ۱. تابع معمولی (همگام / Synchronous)
def normal_func():
    print("تابع معمولی اجرا شد!")
    return 10

# ۲. تابع ناهمگام (Coroutine Function)
async def async_func():
    print("تابع ناهمگام اجرا شد!")
    return 20
```

```python
import asyncio
import time

async def task_one():
    print("[تسک ۱] شروع شد.")
    # کنترل به Event Loop واگذار می‌شود؛ این تسک ۲ ثانیه می‌خوابد
    await asyncio.sleep(2)
    print("[تسک ۱] بعد از ۲ ثانیه بیدار شد و پایان یافت.")
    return "نتیجه تسک ۱"

async def task_two():
    print("[تسک ۲] در حال اجراست...")
    # این تسک هم ۱ ثانیه منتظر می‌ماند
    await asyncio.sleep(1)
    print("[تسک ۲] بعد از ۱ ثانیه کارش تمام شد.")
    return "نتیجه تسک ۲"

async def main():
    start = time.time()
    
    # ساختن Task باعث می‌شود تسک‌ها بلافاصله در Event Loop ثبت شوند
    t1 = asyncio.create_task(task_one())
    t2 = asyncio.create_task(task_two())
    
    # حالا منتظر تمام شدن هر دو می‌مانیم
    await t1
    await t2
    
    print(f"زمان کل: {time.time() - start:.2f} ثانیه")

# اجرا در نوت‌بوک:
await main()

# اجرا در فایل .py:
# asyncio.run(main())
```

```python
import asyncio

async def download_file(file_name, delay):
    print(f"[START] Downloading {file_name}")
    await asyncio.sleep(delay)
    print(f"[DONE] Downloaded {file_name}!")
    return f"{file_name} content"

async def main():
    # Schedule both tasks on the event loop immediately
    task_one = asyncio.create_task(download_file("file_A.zip", 3))
    task_two = asyncio.create_task(download_file("file_B.zip", 1))

    # Await the tasks and capture their return values
    result_one = await task_one
    result_two = await task_two

    # Print the resolved contents
    print("Result 1:", result_one)
    print("Result 2:", result_two)

# Run inside Jupyter / IPython:
await main()
```

```python
import asyncio

async def heartbeat(interval, count):
    for step in range(1, count + 1):
        print(f"[HEARTBEAT] Pulse {step}/{count}: System alive...")
        await asyncio.sleep(interval)
    return "Heartbeat finished"

async def heavy_io_job(job_name, total_duration):
    print(f"[JOB] Starting {job_name} (will take {total_duration}s)...")
    await asyncio.sleep(total_duration)
    print(f"[JOB] Completed {job_name}!")
    return f"{job_name} success"

async def main():
    # Schedule both tasks on the event loop concurrently
    task1 = asyncio.create_task(heartbeat(interval=0.5, count=4))
    task2 = asyncio.create_task(heavy_io_job(job_name="Database Query", total_duration=2.0))

    # Await their completion
    result1 = await task1
    result2 = await task2

    print("Task 1 Result:", result1)
    print("Task 2 Result:", result2)

await main()
```

```python
import asyncio
import time

async def say_hello():
    print("say_hello started...")
    await asyncio.sleep(5)
    return "Hello Async!"

async def say_a_message(message: str):
    print("say_a_message started...")
    await asyncio.sleep(1)
    print(f"The message is -> {message}")

async def main():
    start_time = time.time()

    # 1. Schedule both tasks on the Event Loop immediately
    task1 = asyncio.create_task(say_hello())
    task2 = asyncio.create_task(say_a_message("ehsano"))

    # 2. Wait for both tasks to finish
    msg = await task1
    await task2

    print(msg)
    print(f"Total elapsed time: {time.time() - start_time:.2f}s")

await main()
```

```python
import asyncio
import time

async def blocking_task():
    print("[BLOCKING] Started (using time.sleep)...")
    time.sleep(3)  # Freezes the entire Event Loop for 3 seconds!
    print("[BLOCKING] Finished!")

async def fast_task():
    print("[FAST] Started (using asyncio.sleep)...")
    await asyncio.sleep(1)
    print("[FAST] Finished!")

async def main():
    start_time = time.time()

    # Schedule both tasks concurrently
    t1 = asyncio.create_task(blocking_task())
    t2 = asyncio.create_task(fast_task())

    await t1
    await t2

    print(f"Total elapsed time: {time.time() - start_time:.2f}s")

# Run in Jupyter Notebook:
await main()
```

```python
import asyncio
import time

async def blocking_task():
    print("[BLOCKING] Task started...")
    time.sleep(3)  # Freezes the entire Event Loop for 3 seconds
    print("[BLOCKING] Task finished!")

async def non_blocking_task():
    print("[NON-BLOCKING] Task started, going to sleep for 1s...")
    await asyncio.sleep(1)  # Yields control back to the Event Loop
    print("[NON-BLOCKING] Task woke up and finished!")

async def main():
    total_start = time.time()

    # Schedule both tasks on the Event Loop
    t1 = asyncio.create_task(blocking_task())
    t2 = asyncio.create_task(non_blocking_task())

    # Wait for both tasks to complete
    await t1
    await t2

    total_elapsed = time.time() - total_start
    print(f"Total elapsed time: {total_elapsed:.2f}s")

# Run in Jupyter Notebook:
await main()
```

```python
import asyncio
import time

async def live_clock(ticks):
    for i in range(1, ticks + 1):
        print(f"[CLOCK] Tick {i}/{ticks}")
        await asyncio.sleep(0.5)

async def blocking_api_request():
    print("[API - BLOCKING] Request sent (freezing loop for 2s)...")
    time.sleep(2)  # Blocks the entire thread!
    print("[API - BLOCKING] Response received!")

async def main_blocking():
    start_time = time.time()

    clock_task = asyncio.create_task(live_clock(5))
    api_task = asyncio.create_task(blocking_api_request())

    await clock_task
    await api_task

    print(f"Total time (Blocking): {time.time() - start_time:.2f}s\n")

# Run in Jupyter Notebook:
await main_blocking()
```
